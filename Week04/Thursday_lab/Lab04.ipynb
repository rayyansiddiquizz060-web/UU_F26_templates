Question 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
K = 4

values = np.arange(1, K+1)

devalues = values / K

devalues
```

```python
random_values = np.random.choice(devalues, size=5000)

random_values

x = np.sort(random_values)


y = np.arange(1, len(x) + 1) / len(x)
y 
```

```python
plt.step(x,y)
plt.xlabel("Value")
plt.ylabel("ECDF")
plt.show()
```

As I increase K, the distribution would show more of the steps because there would be more values and have it increase at those values.

Question 2

```python
import numpy as np

n = 32
b= 5000
results = []

for i in range(b) :
    randomNumbers = np.random.uniform(-np.sqrt(3), np.sqrt(3), size=n)

    randomNumbers

    mean = np.mean(randomNumbers)
    mean2 = mean * np.sqrt(n)
    results.append(mean2)
results
    
```

```python
import seaborn as sns

sns.ecdfplot(results)
```

The distribution this plot approaches as n increases is a normal distribution

Question 3

```python
n = 40
p = .50
b = 5000
results = []
for i in range(b) : 
    heads = np.random.binomial(n, p)
    heads1 = (heads - n*p) / np.sqrt(n*p*(1-p))
    results.append(heads1)
results


```

```python
sns.ecdfplot(results)
```

The distribution this plot approaches as n increases is also the normal distribution.
The results change as I adjust p by changing the spread of the results.

Question 4

```python
n = 5000
r = 2
dt = .001
times = []

for i in range(n):
    per = np.random.geometric(r * dt)
    timePeriod = per * dt
    times.append(timePeriod)
times
```

```python
sns.ecdfplot(times)
```

As dt goes to zero, the intervals get smaller and smaller which means it is exponential. 
For r, the results vary based on how big it is. If it is big, then the wait times are shorter and vise versa.
